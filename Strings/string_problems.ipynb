### 1) Check if the string is palindrom or not

```python
str = input("Enter string: ")  #enter the string
str = str.lower()  # make all the string lower or upper case
rev_str = reversed(str)

if list(str)==list(rev_str):
    print(f"{str} is palindrom")
else:
    print(f"{str} is not a palindrom")
```

```python
str = input("Enter string: ")  #enter the string
str = str.lower()  # make all the string lower or upper case
rev_str = str[::-1]

if (str)==(rev_str):
    print(f"{str} is palindrom")
else:
    print(f"{str} is not a palindrom")
```

reversed() is just an iterator, it will iterate thru the string but not produce the reversed string, hence we need to join by '' delimiter.

```python
str = "madam"
str = str.lower()  # make all the string lower or upper case
rev_str = ''.join(reversed(str))
print(str)
print(rev_str)
```

### 2) Python Program to Sort Words in Alphabetic Order.

```python
str = "My name is Anthony Gonsalve. I am alone in this world but very close to God."
# break the string into words
words = str.split()
words.sort()
print(words)
for i in words:
    print(i)
```

3) Reverse a string without slicing

```python
str = "Cabbage"
rev_str = ""  # create empty string
for i in str:
    rev_str = i + rev_str # first 'c' will be iterated and sit in rev_str, then 'a' followed by rev_str hecne it will be reversed string.
print(rev_str)
```

4) Reverse a string with slicing

```python
str = "Cabbage"
rev_str = str[::-1]
print(rev_str)
```
