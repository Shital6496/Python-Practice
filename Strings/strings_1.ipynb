5) typecasting/type conversion means changing a value from one data type to another.  
  a) Explicit typecasting - You manually tell Python to convert the type.

```python
age = "25"

age = int(age)

print(age)
print(type(age))
```

b) Implicit typecasting - Python automatically converts the type when it can do so safely.

```python
a = 10       # int
b = 2.5      # float

result = a + b

print(result)
print(type(result))
```

6) Indexing

```python
my_var = "cat and dog"
print(my_var[0:len(my_var)])
print(my_var[:])
```

7) Upper and lower (capital letters)

```python
my_var = "cat and dog"
print(my_var.upper())
```

```python
my_var = "CAT AND DOG"
print(my_var.lower())
```

8) replace letters

```python
my_var = "CAT AND DOG"
print(my_var.replace("A","Z"))
my_var = "CAT AND DOG"
print(my_var.replace("AT","Z"))
```

12. check if the variable is numeric or alphanumeric

```python
x = "Hello"
y = "100"
z = "100abc"
print(x.isnumeric())
print(x.isalnum())
print(x.isalpha())

print(y.isnumeric())
print(y.isalnum())
print(y.isalpha())

print(z.isnumeric())
print(z.isalnum())
print(z.isalpha())
```
